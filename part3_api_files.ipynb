```python
# For API
import requests

url="https://dummyjson.com/products?limit=20"

# Get data from API
response = requests.get(url)

# Convert to dictionary
data = response.json()

# Get product list
products = data["products"]

# Print table
for p in products:
  print(p["id"], "-", p["title"], "-",p["category"], "-", "$"+str(p["price"]), "-", p["rating"])
```

```python
# Filtered products with rating >= 4.5
filtered = []

for p in products:
  if p["rating"] >= 4.5:
    filtered.append(p)

# Sort by price from high to low
filtered.sort(key=lambda x: x["price"], reverse=True)

print("Filtered Products:")

for p in filtered:
  print(p["title"], "-", "$"+str(p["price"]), "-", p["rating"])
```

```python
# Safe divide function

def safe_divide(a, b):
    try:
      return a / b
    except ZeroDivisionError:
      return "Error: Cannot divide by zero"
    except TypeError:
      return "Error: Invalid input types"

# Test
print(safe_divide(10, 2))
print(safe_divide(10, 0))
print(safe_divide("ten", 2))

```

```python
# Safe file read

def read_file_safe(filename):
  try:
    with open(filename, "r") as f:
      data = f.read()
      return data
  except FileNotFoundError:
    print("Error: File not found")
  finally:
    print("File operation attempt complete")

# Test

print(read_file_safe("python_notes.txt"))
# Should work
print(read_file_safe("ghost_notes.txt"))
# Should not work

```

```python
# Input Validation

import requests

while True:
  user_input = input("Enter producy ID (1-100) or quit: ")

  if user_input == "quit":
    break

  if not user_input.isdigit():
    print("Invalid input")
    continue

  pid = int(user_input)

  if pid < 1 or pid > 100:
    print("Out of range")
    continue

  url =f"https://dummyjson.com/products/{pid}"

  response = requests.get(url)

  if response.status_code == 200:
    data = response.json()
    print(data["title"], "-", data["price"])
  else:
    print("Product not found")
```
